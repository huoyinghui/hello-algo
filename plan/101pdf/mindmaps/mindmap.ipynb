```python
# list 操作
n = 5
# 动态数组
arr = []                    # 创建
arr = [1, 2, 3]            # 初始化
arr = [0] * n              # 固定大小

# 访问 O(1)
val = arr[0]               # 首元素
val = arr[-1]              # 末元素

# 修改 O(1)
arr[0] = 10

# 插入
arr.append(4)              # 末尾 O(1)
arr.insert(0, 0)           # 指定位置 O(n)

# 删除
arr.pop()                  # 末尾 O(1)
arr.pop(0)                 # 指定位置 O(n)
arr.remove(val)            # 删除值 O(n)

# 切片 O(k)
sub = arr[1:3]            # [start:end)
reversed_arr = arr[::-1]   # 反转
```
